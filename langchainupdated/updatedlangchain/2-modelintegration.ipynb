```python
### GROQ Model Initialization
import os
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:qwen/qwen3.8-27b")
response = model.invoke("What does the company Groq do?")
response.content
```

```python
from langchain_groq import ChatGroq
ChatGroq(model="qwen/qwen3.8-27b")
reponse=model.invoke("Can you tell me what does Groq do?")
reponse.content
```

```python
model.invoke("Write me a 200 words paragraph on Artificial Intelligence")
```

```python
for chunk in model.stream("Write me a 200 words paragraph on Artificial Intelligence"):
    print(chunk.text, end="|", flush=True)
```

```python
for chunk in model.stream("Why do parrots have colorful feathers.Give me a detail message"):
    print(chunk.text, flush=True)
```

```python


responses = model.batch(
    [
        "Why do parrots have colourful feathers?",
        "How do birds fly?",
        "What is quantum computing?",
    ],
    config={"max_concurrency": 1},   # run one at a time
)

for r in responses:
    print(r.content)
    print("-" * 40)
```
